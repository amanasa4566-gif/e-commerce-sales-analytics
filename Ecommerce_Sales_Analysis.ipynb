```python
rfm = df_customer.groupby('CustomerID').agg(
    LastPurchaseDate=('InvoiceDate', 'max'),
    Frequency=('InvoiceNo', 'nunique'),
    Monetary=('Revenue', 'sum')
)

rfm.head()
```

```python
reference_date = df_customer['InvoiceDate'].max() + pd.Timedelta(days=1)

rfm['Recency'] = (
    reference_date - rfm['LastPurchaseDate']
).dt.days

rfm.head()
```

```python
rfm[['Recency', 'Frequency', 'Monetary']].describe()
```

```python
rfm['R_Score'] = pd.qcut(
    rfm['Recency'].rank(method='first'),
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

rfm['F_Score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm['M_Score'] = pd.qcut(
    rfm['Monetary'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm.head()
```

```python
rfm['RFM_Score'] = (
    rfm['R_Score'] +
    rfm['F_Score'] +
    rfm['M_Score']
)

rfm[['Recency', 'Frequency', 'Monetary',
     'R_Score', 'F_Score', 'M_Score', 'RFM_Score']].head(10)
```

```python
rfm['RFM_Score'].value_counts().sort_index()
```

```python
def rfm_segment(score):
    if score >= 13:
        return 'Champions'
    elif score >= 10:
        return 'Loyal / Potential Loyalists'
    elif score >= 7:
        return 'Needs Attention'
    elif score >= 4:
        return 'At Risk'
    else:
        return 'Lost'


rfm['Segment'] = rfm['RFM_Score'].apply(rfm_segment)

rfm[['RFM_Score', 'Segment']].head(10)
```

```python
segment_counts = (
    rfm['Segment']
    .value_counts()
)

segment_counts
```

```python
segment_revenue = (
    rfm.groupby('Segment')['Monetary']
    .sum()
    .sort_values(ascending=False)
)

segment_revenue
```

```python
segment_revenue_pct = (
    segment_revenue / segment_revenue.sum() * 100
).round(2)

segment_revenue_pct
```
